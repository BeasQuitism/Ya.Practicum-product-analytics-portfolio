### Приоритизация анализа

**Критически важные метрики (влияют на стратегические решения):**

1. **Retention Rate** — если пользователи не возвращаются, продукт нежизнеспособен
2. **Конверсия в заказ (CR)** — главный индикатор product-market fit
3. **LTV/CAC Ratio** — определяет устойчивость бизнес-модели (цель > 3.0)
4. **Payback Period** — показывает скорость возврата инвестиций (цель < 12 мес.)
5. **Динамика выручки** — индикатор роста или стагнации

**Вспомогательные индикаторы (помогают найти точки роста):**

1. **ARPU/ARPPU/AOV** — диагностические метрики для понимания монетизации
2. **Конверсия по каналам** — для оптимизации маркетингового бюджета
3. **Сегментация** — для персонализации и таргетирования
4. **Устройства/регионы/категории** — для точечных улучшений UX и ассортимента

### Основные термины

**Метрики аудитории:**
- **MAU (Monthly Active Users)** — уникальные пользователи за месяц
- **WAU (Weekly Active Users)** — уникальные пользователи за неделю
- **DAU (Daily Active Users)** — уникальные пользователи за день
- **Stickiness** — отношение DAU к MAU, показывает частоту использования продукта

**Метрики монетизации:**
- **Revenue** — выручка компании (в нашем случае 5% комиссии от GMV)
- **GMV (Gross Merchandise Value)** — общий объем продаж на маркетплейсе
- **ARPU (Average Revenue Per User)** — средняя выручка на одного активного пользователя
- **ARPPU (Average Revenue Per Paying User)** — средняя выручка на одного платящего пользователя
- **AOV (Average Order Value)** — средний чек заказа
- **CR (Conversion Rate)** — конверсия из активного пользователя в покупателя

**Метрики удержания:**
- **Retention Rate** — процент пользователей, которые вернулись и совершили действие через определенный период
- **LTV (Lifetime Value)** — пожизненная ценность клиента (общая выручка от пользователя за все время)
- **Churn Rate** — процент пользователей, которые перестали пользоваться продуктом

**Маркетинговые метрики:**
- **CAC (Customer Acquisition Cost)** — стоимость привлечения одного платящего клиента
- **LTV/CAC Ratio** — соотношение пожизненной ценности к стоимости привлечения (показывает окупаемость)
  * **< 1.0**: убыточная модель
  * **1.0-3.0**: приемлемо, но есть куда расти
  * **> 3.0**: отличная юнит-экономика
- **ROI (Return on Investment)** — возврат инвестиций в процентах
  - Формула: ((LTV - CAC) / CAC) × 100%
- **Payback Period** — срок окупаемости затрат на привлечение (через сколько месяцев LTV покрывает CAC)

**Воронка:**
- **Page View** — просмотр страницы
- **Product View** — просмотр товара
- **Add to Cart** — добавление в корзину
- **Checkout Start** — начало оформления заказа
- **Checkout Complete** — завершение оформления заказа

### 2. Построить обзор ключевых метрик

````python
# Выручка бизнеса = 5% от total_price
orders['revenue'] = orders['total_price'] * 0.05
````

````python
# MAU
mau = events.groupby(events['event_date'].dt.to_period('M'))['user_id'].nunique().reset_index()
mau.columns = ['month', 'mau']

# WAU
wau = events.groupby(events['event_date'].dt.to_period('W'))['user_id'].nunique().reset_index()
wau.columns = ['week', 'wau']

# DAU
dau = events.groupby(events['event_date'].dt.to_period('D'))['user_id'].nunique().reset_index()
dau.columns = ['date', 'dau']

# DAU_7D (Скользящее ЗА 7 ДНЕЙ)
dau['dau_rolling_7'] = dau['dau'].rolling(window=7, min_periods=1).mean()

print(f"Средний MAU: {mau['mau'].mean():.0f} пользователей")
print(f"Средний WAU: {wau['wau'].mean():.0f} пользователей")
print(f"Средний DAU: {dau['dau'].mean():.0f} пользователей")
print(f"Stickiness (DAU/MAU): {(dau['dau'].mean() / mau['mau'].mean() * 100):.2f}%")
````

````python
# Revenue по месяцам
monthly_revenue = orders.groupby(orders['order_date'].dt.to_period('M')).agg({
    'revenue': 'sum',
    'order_id': 'nunique',
    'user_id': 'nunique'
}).reset_index()
monthly_revenue.columns = ['month', 'revenue', 'orders_count', 'buyers_count']

# ARPU
monthly_arpu = monthly_revenue.merge(mau, on='month')
monthly_arpu['arpu'] = monthly_arpu['revenue'] / monthly_arpu['mau']

# ARPPU
monthly_arpu['arppu'] = monthly_arpu['revenue'] / monthly_arpu['buyers_count']

# AOV
monthly_arpu['aov'] = monthly_arpu['revenue'] / monthly_arpu['orders_count']

print(f"Общая выручка: {orders['revenue'].sum():,.2f} руб.")
print(f"Средний ARPU: {monthly_arpu['arpu'].mean():.2f} руб.")
print(f"Средний ARPPU: {monthly_arpu['arppu'].mean():.2f} руб.")
print(f"Средний AOV: {monthly_arpu['aov'].mean():.2f} руб.")
````

````python
# CR в заказ
monthly_cr = events.groupby(events['event_date'].dt.to_period('M'))['user_id'].nunique().reset_index()
monthly_cr.columns = ['month', 'active_users']

monthly_buyers = orders.groupby(orders['order_date'].dt.to_period('M'))['user_id'].nunique().reset_index()
monthly_buyers.columns = ['month', 'buyers']

monthly_cr = monthly_cr.merge(monthly_buyers, on='month', how='left')
monthly_cr['cr_to_order'] = (monthly_cr['buyers'] / monthly_cr['active_users'] * 100).fillna(0)

# Воронка событий
funnel_events = ['page_view', 'product_view', 'add_to_cart', 'checkout_complete']
funnel = events[events['event_type'].isin(funnel_events)].groupby('event_type')['user_id'].nunique().sort_values(ascending=False)

print(f"Средний CR в заказ: {monthly_cr['cr_to_order'].mean():.2f}%")
print(funnel)
````

````python
# Когортный анализ по месяцам
def calculate_retention(users_df, orders_df):
    # Первая покупка каждого пользователя
    first_purchase = orders_df.groupby('user_id')['order_date'].min().reset_index()
    first_purchase.columns = ['user_id', 'first_purchase_date']
    
    # Когорта пользователя (месяц первой покупки)
    first_purchase['cohort_month'] = first_purchase['first_purchase_date'].dt.to_period('M')
    
    # Объединяем с заказами
    orders_with_cohort = orders_df.merge(first_purchase[['user_id', 'cohort_month']], on='user_id')
    
    # Номер месяца после первой покупки
    orders_with_cohort['month_number'] = (
        orders_with_cohort['order_date'].dt.to_period('M') - 
        orders_with_cohort['cohort_month']
    ).apply(lambda x: x.n)
    
    # Размер когорт
    cohort_sizes = first_purchase.groupby('cohort_month')['user_id'].nunique().reset_index()
    cohort_sizes.columns = ['cohort_month', 'cohort_size']
    
    # Количество активных пользователей по когортам и месяцам
    retention_data = orders_with_cohort.groupby(['cohort_month', 'month_number'])['user_id'].nunique().reset_index()
    retention_data = retention_data.merge(cohort_sizes, on='cohort_month')
    
    # Retention rate
    retention_data['retention_rate'] = (retention_data['user_id'] / retention_data['cohort_size'] * 100).round(2)
    
    return retention_data

retention_df = calculate_retention(users, orders)

retention_pivot = retention_df.pivot(index='cohort_month', columns='month_number', values='retention_rate')
print(retention_pivot.head())

# Строим график
plt.figure(figsize=(12, 8))

sns.heatmap(
    retention_pivot,
    annot=True,
    fmt='.1f',
    cmap='coolwarm',
    cbar_kws={'label': 'Retention Rate (%)'})

# Заголовки и подписи
plt.title('Retention Rate по когортам пользователей (%)', 
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Месяцев после первой покупки', fontsize=12, labelpad=10)
plt.ylabel('Когорта (месяц первой покупки)', fontsize=12, labelpad=10)
plt.yticks(rotation=0)

plt.tight_layout()

plt.show()
````

````python
# CAC
monthly_cac = campaign_costs.groupby(campaign_costs['spend_month'].dt.to_period('M'))['budget'].sum().reset_index()
monthly_cac.columns = ['month', 'marketing_spend']

# Новые пользователи по месяцам
new_users = users.groupby(users['registration_date'].dt.to_period('M'))['user_id'].nunique().reset_index()
new_users.columns = ['month', 'new_users']

monthly_cac = monthly_cac.merge(new_users, on='month', how='left')
monthly_cac['cac'] = monthly_cac['marketing_spend'] / monthly_cac['new_users']

# LTV/CAC ratio
ltv_cac = monthly_arpu.merge(monthly_cac, on='month', how='inner')
ltv_cac['ltv_cac_ratio'] = ltv_cac['arppu'] / ltv_cac['cac']

print(f"Общие маркетинговые затраты: {campaign_costs['budget'].sum():,.2f} руб.")
print(f"Средний CAC: {monthly_cac['cac'].mean():.2f} руб.")
print(f"Средний LTV/CAC ratio: {ltv_cac['ltv_cac_ratio'].mean():.2f}")
````

````python
# LTV
def calculate_cohort_ltv(orders_df, users_df):
    # Первая покупка каждого пользователя
    first_order = orders_df.groupby('user_id')['order_date'].min().reset_index()
    first_order.columns = ['user_id', 'first_order_date']
    first_order['cohort_month'] = first_order['first_order_date'].dt.to_period('M')
    
    # Все заказы с когортой
    orders_with_cohort = orders_df.merge(
        first_order[['user_id', 'cohort_month']], 
        on='user_id'
    )
    
    # Месяц от первой покупки
    orders_with_cohort['months_since_first'] = (
        orders_with_cohort['order_date'].dt.to_period('M') - 
        orders_with_cohort['cohort_month']
    ).apply(lambda x: x.n)
    
    # LTV по когортам (накопленная выручка)
    cohort_ltv = orders_with_cohort.groupby(['cohort_month', 'months_since_first']).agg({
        'revenue': 'sum',
        'user_id': 'nunique'
    }).reset_index()
    
    # Накопительный итог
    cohort_ltv['cumulative_revenue'] = cohort_ltv.groupby('cohort_month')['revenue'].cumsum()
    cohort_ltv['ltv'] = cohort_ltv['cumulative_revenue'] / cohort_ltv.groupby('cohort_month')['user_id'].transform('first')
    
    return cohort_ltv

cohort_ltv = calculate_cohort_ltv(orders, users)

# Средний LTV за 6 месяцев
ltv_6m = cohort_ltv[cohort_ltv['months_since_first'] <= 6].groupby('cohort_month').agg({
    'ltv': 'max'
}).reset_index()

print(ltv_6m)
print(f"\nСредний LTV (6 мес): {ltv_6m['ltv'].mean():.2f} руб.")
````

````python
# Объединяем LTV и CAC для расчета ROI
unit_economics = ltv_6m.merge(
    monthly_cac[['month', 'cac']].rename(columns={'month': 'cohort_month'}), 
    on='cohort_month', 
    how='inner'
)

# ROI = (LTV - CAC) / CAC * 100
unit_economics['roi'] = ((unit_economics['ltv'] - unit_economics['cac']) / unit_economics['cac']) * 100

# LTV/CAC ratio
unit_economics['ltv_cac_ratio'] = unit_economics['ltv'] / unit_economics['cac']


print(f"\nСредний ROI: {unit_economics['roi'].mean():.1f}%")
print(f"Средний LTV/CAC: {unit_economics['ltv_cac_ratio'].mean():.2f}")
````

````python
# Payback Period
def calculate_payback(orders_df, users_df, cac_df):
    # Первая покупка
    first_order = orders_df.groupby('user_id')['order_date'].min().reset_index()
    first_order.columns = ['user_id', 'first_order_date']
    first_order['cohort_month'] = first_order['first_order_date'].dt.to_period('M')
    
    # Все заказы с когортой
    orders_with_cohort = orders_df.merge(first_order, on='user_id')
    orders_with_cohort['order_month'] = orders_with_cohort['order_date'].dt.to_period('M')
    orders_with_cohort['months_since_first'] = (
        orders_with_cohort['order_month'] - orders_with_cohort['cohort_month']
    ).apply(lambda x: x.n)
    
    # Размер когорт 
    cohort_sizes = first_order.groupby('cohort_month')['user_id'].nunique().reset_index()
    cohort_sizes.columns = ['cohort_month', 'cohort_size']
    
    # Выручка по когортам и месяцам
    cohort_revenue = orders_with_cohort.groupby(['cohort_month', 'months_since_first']).agg({
        'revenue': 'sum',
        'user_id': 'nunique' 
    }).reset_index()
    
    # Объединяем с размером когорт
    cohort_revenue = cohort_revenue.merge(cohort_sizes, on='cohort_month')
    
    # Выручка на пользователя когорты
    cohort_revenue['revenue_per_user'] = (
        cohort_revenue['revenue'] / cohort_revenue['cohort_size']
    )
    
    # Считаем payback
    payback_results = []
    
    for cohort in sorted(cohort_revenue['cohort_month'].unique()):
        cohort_data = cohort_revenue[
            cohort_revenue['cohort_month'] == cohort
        ].sort_values('months_since_first')
        
        if len(cohort_data) == 0:
            continue
        
        # Находим CAC
        cac_row = cac_df[cac_df['month'] == cohort]
        if len(cac_row) == 0:
            continue
        cac = cac_row['cac'].values[0]
        
        # Накопительный LTV
        cumulative_ltv = 0
        payback_month = None
        
        for _, row in cohort_data.iterrows():
            month_num = int(row['months_since_first'])
            monthly_revenue = row['revenue_per_user']
            
            cumulative_ltv += monthly_revenue
            
            # Проверяем окупаемость
            if cumulative_ltv >= cac and payback_month is None:
                payback_month = month_num
        
        payback_results.append({
            'cohort_month': str(cohort),
            'cac': round(cac, 2),
            'ltv_total': round(cumulative_ltv, 2),
            'payback_months': payback_month
        })
    
    return pd.DataFrame(payback_results)

payback_df = calculate_payback(orders, users, monthly_cac)

print(payback_df)
print(f"\nСредний Payback Period: {payback_df['payback_months'].mean():.1f} мес.")
````

````python
# Основные метрики по каналам
channel_metrics = users.groupby('acq_channel').agg({
    'user_id': 'nunique'
}).reset_index()
channel_metrics.columns = ['channel', 'total_users']

# Выручка по каналам
channel_revenue = orders.merge(users[['user_id', 'acq_channel']], on='user_id')
channel_revenue = channel_revenue.groupby('acq_channel').agg({
    'revenue': 'sum',
    'order_id': 'nunique',
    'user_id': 'nunique',
    'total_price': 'sum'
}).reset_index()
channel_revenue.columns = ['channel', 'revenue', 'orders', 'buyers', 'gmv']

channel_analysis = channel_metrics.merge(channel_revenue, on='channel', how='left')

# ARPU, ARPPU, AOV, CR
channel_analysis['arpu'] = channel_analysis['revenue'] / channel_analysis['total_users']
channel_analysis['arppu'] = channel_analysis['revenue'] / channel_analysis['buyers']
channel_analysis['aov'] = channel_analysis['revenue'] / channel_analysis['orders']
channel_analysis['conversion_rate'] = (channel_analysis['buyers'] / channel_analysis['total_users'] * 100)

# CAC по каналам
channel_cac = campaign_costs.groupby('acq_channel')['budget'].sum().reset_index()
channel_cac.columns = ['channel', 'total_spend']

channel_analysis = channel_analysis.merge(channel_cac, on='channel', how='left')
channel_analysis['cac'] = channel_analysis['total_spend'] / channel_analysis['total_users']

# LTV
channel_analysis['ltv'] = channel_analysis['revenue'] / channel_analysis['total_users']
channel_analysis['ltv_cac_ratio'] = channel_analysis['ltv'] / channel_analysis['cac']

# Сортируем по выручке
channel_analysis = channel_analysis.sort_values('revenue', ascending=False)

print(channel_analysis[['channel', 'total_users', 'revenue', 'arpu', 'arppu', 'aov', 
                        'conversion_rate', 'cac', 'ltv_cac_ratio']].to_string(index=False))
````

# Дизайн эксперимента:

## 1. Контекст и цель эксперимента

- **Бизнес-контекст:** Команда продукта запускает эксперимент, направленный на повышение конверсии среди пользователей, привлечённых через канал TikTok. Для стимулирования первой покупки пользователям будет предоставлен бонус на первую покупку.  
- **Цель:** Проверить, влияет ли предоставление бонуса на первую покупку на конверсию и выручку среди пользователей TikTok.  
- **Гипотеза:** Предоставление бонуса на первую покупку новым пользователям, привлеченным через TikTok, приведет к увеличению конверсии в первую покупку

---

## 2. Формат эксперимента

- **Тип эксперимента:** Сплит-роллинг (rolling split)  
- **Канал:** Только пользователи, пришедшие с TikTok  
- **Группы:**  
  - **Контрольная (A):** пользователи TikTok без бонуса  
  - **Тестовая (B):** пользователи TikTok с бонусом на первую покупку  
- **Распределение:** Равномерное распределение новых пользователей TikTok между группами по дате регистрации  
- **Период проведения:** 3 недели (фиксировано)

---

## 3. Ключевые метрики

### Основная метрика

- **Конверсия в первую покупку** — доля пользователей, зарегистрировавшихся с TikTok и совершивших первую покупку  
  - *Цель:* увеличить конверсию с помощью бонуса  
  - *Критерий успеха:* статистически значимый рост конверсии в тестовой группе

### Вспомогательные метрики

- **ARPU** (средняя выручка на пользователя TikTok)  
- **ARPPU** (средняя выручка на покупателя TikTok)  
- **AOV** (средний чек TikTok)

### Барьерная метрика

- **Общая выручка от пользователей TikTok**

---

## 4. Размеры групп и запуск

- **Целевая аудитория:** только пользователи TikTok  
- **Размеры групп:** определяются по фактическому трафику TikTok за период эксперимента  
- **Комментарий:** запуск без предварительного расчёта мощности и MDE из-за срочности задачи

````python
# Расчитываем метрики

# Конверсия в первую покупку - основная метрика
users_with_group = users_ab.merge(ab_split, on='user_id', how='inner')

# Первая покупка каждого пользователя
first_purchase = orders_ab.groupby('user_id')['order_date'].min().reset_index()
first_purchase.columns = ['user_id', 'first_purchase_date']

# Объединяем с пользователями
users_with_purchase = users_with_group.merge(first_purchase, on='user_id', how='left')
users_with_purchase['has_purchase'] = users_with_purchase['first_purchase_date'].notna()

# Конверсия по группам
conversion_by_group = users_with_purchase.groupby('group').agg({
    'user_id': 'count',
    'has_purchase': 'sum'
}).reset_index()
conversion_by_group.columns = ['group', 'total_users', 'purchasers']
conversion_by_group['conversion_rate'] = round((
    conversion_by_group['purchasers'] / conversion_by_group['total_users'] * 100), 2
)

print(conversion_by_group)
````

````python
# Финансовые метрики
# Выручка = 5% от total_price
orders_ab['revenue'] = orders_ab['total_price'] * 0.05

# Выручка по группам - Барьерная метрика
orders_with_group = orders_ab.merge(ab_split, on='user_id', how='inner')
revenue_by_group = orders_with_group.groupby('group').agg({
    'revenue': 'sum',
    'order_id': 'count',
    'user_id': 'nunique',
    'total_price': 'mean'
}).reset_index().round(2)
revenue_by_group.columns = ['group', 'total_revenue', 'orders_count', 'buyers', 'aov']

# Объединяем с пользователями для ARPU
revenue_by_group = revenue_by_group.merge(
    conversion_by_group[['group', 'total_users']], 
    on='group'
)
revenue_by_group['arpu'] = round(revenue_by_group['total_revenue'] / revenue_by_group['total_users'], 2)
revenue_by_group['arppu'] = round(revenue_by_group['total_revenue'] / revenue_by_group['buyers'], 2)

print(revenue_by_group)
````

````python
# Расчет разницы между группами
group_ctrl = conversion_by_group[conversion_by_group['group'] == 'control'].iloc[0]
group_trmnt = conversion_by_group[conversion_by_group['group'] == 'treatment'].iloc[0]

change_cr = (group_trmnt ['conversion_rate'] - group_ctrl ['conversion_rate']) / group_ctrl ['conversion_rate'] * 100
print(f"Разница конверсии: {change_cr:.2f}%")

rev_ctrl = revenue_by_group[revenue_by_group['group'] == 'control'].iloc[0]
rev_trmnt = revenue_by_group[revenue_by_group['group'] == 'treatment'].iloc[0]
change_arpu= (rev_trmnt['arpu'] - rev_ctrl['arpu']) / rev_ctrl['arpu'] * 100
print(f"Разница ARPU: {change_arpu:.2f}%")
````

````python
# Статистическая значимость
# Z-test для конверсии (пропорции)
n_ctrl = group_ctrl['total_users']
n_trmnt = group_trmnt['total_users']
x_ctrl = group_ctrl['purchasers']
x_trmnt = group_trmnt['purchasers']

# Z-test для двух пропорций
count = np.array([x_trmnt, x_ctrl])
nobs = np.array([n_trmnt, n_ctrl])
stat, p_value_cr = proportions_ztest(count, nobs)

print(f"Конверсия control: {group_ctrl['conversion_rate']:.2f}%")
print(f"Конверсия treatment: {group_trmnt['conversion_rate']:.2f}%")
print(f"Z-statistic: {stat:.4f}")
print(f"P-value: {p_value_cr:.4f}")

if p_value_cr < 0.05:
    print("Разница статистически значима (p < 0.05)")
else:
    print("Разница статистически НЕ значима (p >= 0.05)")
````

````python
# T-test ARPU

# Выручка на пользователя
revenue_per_user_ctrl = orders_with_group[orders_with_group['group'] == 'control'].groupby('user_id')['revenue'].sum()
revenue_per_user_trmnt = orders_with_group[orders_with_group['group'] == 'treatment'].groupby('user_id')['revenue'].sum()

# Добавляем нули для тех, кто не покупал
all_users_ctrl = users_with_group[users_with_group['group'] == 'control']['user_id']
all_users_trmnt = users_with_group[users_with_group['group'] == 'treatment']['user_id']

revenue_per_user_ctrl = revenue_per_user_ctrl.reindex(all_users_ctrl, fill_value=0)
revenue_per_user_trmnt = revenue_per_user_trmnt.reindex(all_users_trmnt, fill_value=0)

# T-test (непарный)
t_stat, p_value_arpu = stats.ttest_ind(revenue_per_user_trmnt, revenue_per_user_ctrl, equal_var=False)

print(f"ARPU control: {revenue_per_user_ctrl.mean():.2f} руб.")
print(f"ARPU treatment: {revenue_per_user_trmnt.mean():.2f} руб.")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value_arpu:.4f}")

if p_value_arpu < 0.05:
    print("Разница статистически значима (p < 0.05)")
else:
    print("Разница статистически НЕ значима (p >= 0.05)")
````

````python
# T-test AOV
aov_ctrl = orders_with_group[orders_with_group['group'] == 'control']['total_price']
aov_trmnt = orders_with_group[orders_with_group['group'] == 'treatment']['total_price']

t_stat_aov, p_value_aov = stats.ttest_ind(aov_trmnt, aov_ctrl, equal_var=False)

print(f"AOV control: {aov_ctrl.mean():.2f} руб.")
print(f"AOV treatment: {aov_trmnt.mean():.2f} руб.")
print(f"T-statistic: {t_stat_aov:.4f}")
print(f"P-value: {p_value_aov:.4f}")

if p_value_aov < 0.05:
    print("Разница статистически значима (p < 0.05)")
else:
    print("Разница статистически НЕ значима (p >= 0.05)")
````
